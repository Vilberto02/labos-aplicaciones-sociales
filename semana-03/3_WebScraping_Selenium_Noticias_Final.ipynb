### Ejecutando el Script y Generando el dataset (csv / excel)

```python
#creamos el DataFrame
#df = pd.DataFrame(columns=['Column1', 'Column2', 'Column3'])
df = pd.DataFrame(columns=['N','TITULO', 'LINK', 'AUTOR', 'FECHA', 'CONTENIDO', 'CAPTURA'])

#codigo final a ejecutar 
driver.get('https://larepublica.pe/economia')
listado = driver.find_elements(By.CSS_SELECTOR, "div[class='ListSection_list__section--item__zeP_z extend-link--outside']")
#recorrer noticia a noticia
for indice, noticia in enumerate(listado):
    print("*****************************************************")
    print(noticia.text)
    #recuperar todos los campos
    titulo = noticia.find_elements(By.CSS_SELECTOR, "a")[0].text
    link = noticia.find_elements(By.CSS_SELECTOR, "a")[0].get_attribute('href')
    autor = noticia.find_element(By.CSS_SELECTOR, "span[class='AuthorSign_left__TQYRf AuthorSign_authorSign__name__FXLMu']").text
    fecha = noticia.find_element(By.CSS_SELECTOR, "time").text
    #cargamos un segundo navegador para leer la noticia completa
    driver2 = webdriver.Chrome(service=service, options=options)
    driver2.get(link)
    driver2.execute_script("window.scrollTo(50, window.innerHeight);")
    time.sleep(3)
    try:
        contenido = driver2.find_element(By.CSS_SELECTOR, "div[class='MainContent_main__body__i6gEa']").text
        nombre_captura = 'captura'+ str(indice) +'.png'
        driver2.save_screenshot(nombre_captura)
        driver2.close()
    except:
        contenido = "SIN CONTENIDO"
        nombre_captura = '-'
    #***********************************
    # se va agregando al DataFrame
    row = {'N': indice, 
           'TITULO': titulo, 
           'LINK': link,
           'AUTOR': autor,
           'FECHA': fecha,
           'CONTENIDO': contenido,
           'CAPTURA': nombre_captura
          }
    #df = df.append(row, ignore_index = True)
    df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)

driver.close()
```
